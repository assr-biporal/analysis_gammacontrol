```python
import collections
import json
import multiprocessing
import pathlib

import numpy
import matplotlib
import matplotlib.pyplot as plt
import mne
```

```python
file_path = pathlib.Path.cwd()/'passive1.bdf'
assert file_path.exists()
```

```python
raw = mne.io.read_raw_bdf(file_path, exclude=['EXG8'], preload=True)
raw.filter(1, 100)
raw.notch_filter(60)
EXGchannels_dict = dict(
    EXG1="misc",
    EXG2="misc",
    EXG3="eog",
    EXG4="eog",
    EXG5="eog",
    EXG6="misc",
    EXG7="misc"
)
raw.set_channel_types(EXGchannels_dict)
biosemi_montage = mne.channels.make_standard_montage('biosemi64')
raw.set_montage(biosemi_montage)
```
